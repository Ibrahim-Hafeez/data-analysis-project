```python
import pandas as pd
import numpy as np
from sklearn.impute import KNNImputer

URL = 'https://raw.githubusercontent.com/datasciencedojo/datasets/master/titanic.csv'
df = pd.read_csv(URL)

print(f"Dataset loaded: {df.shape[0]} rows, {df.shape[1]} columns")
print("\nMissing values in our three target columns:")
print(df[['Age', 'Embarked', 'Cabin']].isna().sum())
```

```python
# What would naive imputation do?
overall_mean = df['Age'].mean()
print(f"Overall mean Age (naive fill value): {overall_mean:.1f} years")

print("\nActual mean Age within each passenger class:")
print(df.groupby('Pclass')['Age'].mean().round(1))
```

```python
# Step 1: Look at the distribution first
print("Embarked value counts:")
print(df['Embarked'].value_counts())

# Step 2: Extract the mode (the most frequent value)
embarked_mode = df['Embarked'].mode()[0]
print(f"\nMode (most common port): '{embarked_mode}'")

print(f"\nMissing Embarked BEFORE fill: {df['Embarked'].isna().sum()}")
df['Embarked'] = df['Embarked'].fillna(embarked_mode)
print(f"Missing Embarked AFTER fill:  {df['Embarked'].isna().sum()}")
```

```python
# See what each class's median age actually is
print("Median Age by passenger class:")
print(df.groupby('Pclass')['Age'].median())

print(f"\nOverall median (what naive fill would use): {df['Age'].median()}")
print(f"Missing Age values we need to fill: {df['Age'].isna().sum()}")
```

```python
# Fill each missing Age with the median from THAT passenger's Pclass group
df['Age'] = df.groupby('Pclass')['Age'].transform(
  lambda x: x.fillna(x.median())
)

print(f"Missing Age AFTER fill: {df['Age'].isna().sum()}")

# Verify: the group structure is preserved after filling
print("\nMean Age by Pclass after filling:")
print(df.groupby('Pclass')['Age'].mean().round(1))
```

```python
# Create a new column: 1 if the passenger had a cabin, 0 if not
df['has_cabin'] = df['Cabin'].notna().astype(int)

print("has_cabin distribution:")
print(df['has_cabin'].value_counts())

print("\nCross-check: has_cabin by Pclass")
print(pd.crosstab(df['Pclass'], df['has_cabin'],
            rownames=['Pclass'], colnames=['has_cabin']))
```

```python
# Load a fresh copy so we can compare fairly
df_fresh = pd.read_csv(URL)

# KNN only works with numbers - select numeric columns
numeric_cols = ['Age', 'Pclass', 'SibSp', 'Parch', 'Fare']
df_subset = df_fresh[numeric_cols].copy()

# Create the imputer - n_neighbors=5 means "find the 5 most similar rows"
imputer = KNNImputer(n_neighbors=5)

# fit_transform: learn from the data, then fill the missing values
df_knn_filled = pd.DataFrame(
  imputer.fit_transform(df_subset),
  columns=numeric_cols
)

print("KNN-imputed Age - mean by Pclass:")
print(df_knn_filled.groupby(df_fresh['Pclass'])['Age'].mean().round(1))

print("\nGroup-conditional Age - mean by Pclass (our earlier method):")
print(df.groupby('Pclass')['Age'].mean().round(1))
```

```python
print("=== Final Missing Value Audit ===")
print(df[['Age', 'Embarked', 'Cabin', 'has_cabin']].isna().sum())

print(f"\nDataset shape: {df.shape}")
print("\nSummary of treatments applied:")
print("  Embarked → mode fill with 'S' (MCAR treatment)")
print("  Age      → group-conditional median by Pclass (MAR treatment)")
print("  Cabin    → converted to has_cabin binary feature (MNAR treatment)")
```

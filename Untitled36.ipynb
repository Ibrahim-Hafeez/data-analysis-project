```python
# ============================================================
# Q1 — Three-table JOIN
# Business question: Which product category generated the highest total revenue in each region?
# ============================================================

q1 = """
WITH category_revenue AS (
    SELECT
        c.region,
        p.category,
        SUM(f.sale_amount) AS total_revenue,
        ROW_NUMBER() OVER (
            PARTITION BY c.region
            ORDER BY SUM(f.sale_amount) DESC
        ) AS rn
    FROM FactSales f
    JOIN DimCustomer c
        ON f.customer_id = c.customer_id
    JOIN DimProduct p
        ON f.product_id = p.product_id
    GROUP BY c.region, p.category
)
SELECT
    region,
    category,
    total_revenue
FROM category_revenue
WHERE rn = 1
ORDER BY region;
"""

print("-- Q1: Which product category generated the highest total revenue in each region?")
print(q1)
print("Q1 OUTPUT:")
print(q(q1).to_string(index=False))


# ============================================================
# Q2 — Conditional Aggregation
# Business question: How much revenue came from sales above ₹50,000 in each product category?
# ============================================================

q2 = """
SELECT
    p.category,
    SUM(f.sale_amount) AS total_revenue,
    SUM(
        CASE
            WHEN f.sale_amount > 50000 THEN f.sale_amount
            ELSE 0
        END
    ) AS revenue_above_50000
FROM FactSales f
JOIN DimProduct p
    ON f.product_id = p.product_id
GROUP BY p.category
ORDER BY revenue_above_50000 DESC;
"""

print("\n-- Q2: How much revenue came from sales above ₹50,000 in each product category?")
print(q2)
print("Q2 OUTPUT:")
print(q(q2).to_string(index=False))


# ============================================================
# Q3 — Subquery in WHERE
# Business question: Which products generated more revenue than the average product revenue?
# ============================================================

q3 = """
SELECT
    p.product_id,
    p.product_name,
    SUM(f.sale_amount) AS total_revenue
FROM FactSales f
JOIN DimProduct p
    ON f.product_id = p.product_id
WHERE p.product_id IN (
    SELECT product_id
    FROM FactSales
    GROUP BY product_id
    HAVING SUM(sale_amount) > (
        SELECT AVG(product_revenue)
        FROM (
            SELECT
                product_id,
                SUM(sale_amount) AS product_revenue
            FROM FactSales
            GROUP BY product_id
        )
    )
)
GROUP BY p.product_id, p.product_name
ORDER BY total_revenue DESC;
"""

print("\n-- Q3: Which products generated more revenue than the average product revenue?")
print(q3)
print("Q3 OUTPUT:")
print(q(q3).to_string(index=False))
```
